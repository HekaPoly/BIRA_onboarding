# Listes et dictionnaires

Les **listes** sont des structures de données qui ressemblent aux tableaux que l’on retrouve dans d’autres langages.
Elles permettent de stocker **plusieurs éléments** dans une seule variable.

Caractéristiques principales :

* Une liste peut contenir **n’importe quel type de variable** (entiers, flottants, chaînes, booléens, voire d’autres listes).
* Une liste peut contenir **autant d’éléments que vous le souhaitez**.
* Les éléments sont **ordonnés** et accessibles grâce à leur **index** (la position, en commençant par 0).

````python
mylist = []
mylist.append(1)
mylist.append(2)
mylist.append(3)
print(mylist[0]) # prints 1
print(mylist[1]) # prints 2
print(mylist[2]) # prints 3

# prints out 1,2,3
for x in mylist:
    print(x)
````

Accéder à un indice qui n'existe pas génère une exception (une erreur).

````python
mylist = [1,2,3]
print(mylist[2])
````

Un **dictionnaire** est une autre structure de données, similaire à un tableau, mais il utilise des **clés** et des **valeurs** au lieu d’indices numériques.

Caractéristiques principales :

* Chaque élément est sous la forme : `clé : valeur`.
* Les clés doivent être **uniques**.
* Les dictionnaires sont très pratiques pour représenter des données structurées.

````python
phonebook = {}
phonebook["John"] = 938477566
phonebook["Jack"] = 938377264
phonebook["Jill"] = 947662781
print(phonebook)
````

Alternativement, un dictionnaire peut être initialisé avec les mêmes valeurs dans la notation suivante :


````python
phonebook = {
    "John" : 938477566,
    "Jack" : 938377264,
    "Jill" : 947662781
}
print(phonebook)
````

# Set et Tuple

### Set
Un **set** est une collection **non ordonnée** d’éléments **uniques**.
Contrairement aux listes, il ne peut pas contenir de doublons : si un élément est ajouté plusieurs fois, il n’apparaît qu’une seule fois.

Les sets sont particulièrement utiles pour effectuer des opérations **ensemblistes**, similaires à celles que l’on utilise en mathématiques :

* **Union** : combiner les éléments de deux ensembles.
* **Intersection** : obtenir uniquement les éléments communs.
* **Différence** : récupérer les éléments présents dans un set mais pas dans l’autre.
* **Différence symétrique** : garder les éléments qui ne sont pas communs.

#### Principales opérations sur les sets

| Fonction / Opération              | Description                            | Exemple           | Résultat  |         |           |
| --------------------------------- | -------------------------------------- | ----------------- | --------- | ------- | --------- |
| `.add(x)`                         | Ajoute un élément                      | `{1,2}.add(3)`    | `{1,2,3}` |         |           |
| `.remove(x)`                      | Supprime un élément (erreur si absent) | `{1,2}.remove(1)` | `{2}`     |         |           |
| `.union(t)` ou \`                 | \`                                     | Union             | \`{1,2}   | {2,3}\` | `{1,2,3}` |
| `.intersection(t)` ou `&`         | Intersection                           | `{1,2} & {2,3}`   | `{2}`     |         |           |
| `.difference(t)` ou `-`           | Différence                             | `{1,2,3} - {2,3}` | `{1}`     |         |           |
| `.symmetric_difference(t)` ou `^` | Différence symétrique                  | `{1,2} ^ {2,3}`   | `{1,3}`   |         |           |



````python
a = set(["Jake", "John", "Eric"])
print(a)
b = set(["John", "Jill"])
print(b)
````

Pour savoir quels membres ont assisté aux deux événements, vous pouvez utiliser la méthode "intersection" :

````python
a = set(["Jake", "John", "Eric"])
b = set(["John", "Jill"])

print(a.intersection(b))
print(b.intersection(a))
````

### Tuple
Un **tuple** est une structure de données qui ressemble à une liste, mais qui est **immuable** :

* Une fois créé, on ne peut **ni ajouter**, **ni supprimer**, ni **modifier** ses éléments.
* Ils sont donc utiles pour représenter des données **fixes** qui ne doivent pas changer, comme des coordonnées (x, y) ou une date (jour, mois, année).


#### Principales opérations sur les tuples

| Opération             | Description                                       | Exemple                      | Résultat        |
| --------------------- | ------------------------------------------------- | ---------------------------- | --------------- |
| Indexation            | Accéder à un élément par son index                | `t = (1,2,3); t[0]`          | `1`             |
| Découpage (slicing)   | Extraire une portion du tuple                     | `t = (1,2,3,4); t[1:3]`      | `(2,3)`         |
| Concaténation         | Combiner deux tuples                              | `(1,2) + (3,4)`              | `(1,2,3,4)`     |
| Répétition            | Répéter les éléments                              | `(1,2) * 3`                  | `(1,2,1,2,1,2)` |
| Appartenance          | Vérifier si un élément est présent                | `3 in (1,2,3)`               | `True`          |
| Longueur              | Nombre d’éléments dans le tuple                   | `len((1,2,3,4))`             | `4`             |



````python
coordonnees = (10, 20)
print(coordonnees[0]) # affiche 10
print (coordonnees [1])  # affiche 20

# coordonnees[0] = 15 # Erreur : les tuples ne peuvent pas être modifiés
````

# Le module `collections` — Container datatypes

Le module **`collections`** implémente des **types de conteneurs spécialisés** qui complètent ou améliorent les conteneurs standards de Python (`dict`, `list`, `set`, `tuple`).

### Principales classes du module `collections`

| Classe                               | Description                                                                     | Exemple d’utilisation                     |
| ------------------------------------ | ------------------------------------------------------------------------------- | ----------------------------------------- |
| `deque`                              | File double (ajout/retrait rapide aux deux extrémités).                         | File d’attente, pile.                     |

## La classe `collections.deque`

Un **deque** (prononcé "deck") signifie *double-ended queue* (file à deux extrémités).
C’est une structure de données qui généralise :

* **la pile (stack)** : LIFO (Last In, First Out).
* **la file (queue)** : FIFO (First In, First Out).

### Caractéristiques

* Ajouts (`append`, `appendleft`) et suppressions (`pop`, `popleft`) possibles **aux deux extrémités**.
* Complexité **O(1)** pour ces opérations, bien plus efficace qu’une liste Python utilisée comme file.
* Taille optionnelle avec l’argument `maxlen`. Quand la limite est atteinte, les nouveaux ajouts remplacent les plus anciens.
* **Thread-safe** : utilisable dans des contextes multi-threads sans risque de corruption.

````python
from collections import deque
d = deque('ghi') # make a new deque with three items
for elem in d: # iterate over the deque's elements
    print(elem.upper())
````

# Accès aux fichiers
En Python, la gestion des fichiers se fait principalement avec la fonction **`open()`**. Elle permet d’ouvrir un fichier dans un certain **mode** (lecture, écriture, ajout, etc.), puis d’interagir avec son contenu.

**Ouvrir un fichier**
La syntaxe de base est :

````python
f = open("fichier.txt", mode="r")
````

### Principaux modes d’ouverture


| Mode   | Signification                                                   | Exemple                    |
| ------ | --------------------------------------------------------------- | -------------------------- |
| `"r"`  | Lecture (erreur si le fichier n’existe pas).                    | `open("data.txt", "r")`    |
| `"w"`  | Écriture (crée un fichier vide ou **écrase** s’il existe déjà). | `open("data.txt", "w")`    |
| `"a"`  | Ajout (écrit à la fin sans effacer le contenu existant).        | `open("data.txt", "a")`    |
| `"b"`  | Mode binaire (souvent combiné à `"rb"` ou `"wb"`).              | `open("image.png", "rb")`  |
| `"x"`  | Création (erreur si le fichier existe déjà).                    | `open("nouveau.txt", "x")` |


### Bonne pratique : utiliser `with`

L’utilisation du mot-clé **`with`** est recommandée : il gère automatiquement la fermeture du fichier, même en cas d’erreur.

````python
with open("fichier.txt", "r") as f:
    contenue = f.read()
    print(contenue)
````

# Fonctions et lambda

### 1. Les fonctions
Une **fonction** permet de regrouper du code réutilisable.
Elle prend éventuellement des **paramètres**, exécute des instructions, puis peut **retourner un résultat**

**Avantages :**
* Évite de répéter du code
* Rend le programme plus clair et structuré
* Permet de tester des parties de code séparément

````python
def addition(a, b):
    return a + b

# Utilisation
resultat = addition(5, 3)
print(resultat)  # 8
````

### 2. Les fonctions lambda
Une **lambda** est une fonction **anonyme** (sans nom).
Elles sont utiles pour des opérations simples et rapides, généralement en **une seule ligne**.


````python
addition = lambda a, b: a + b
print(addition(5, 3))  # 8
````

**Autres exemples pratiques :**
* Avec `map` (appliquer une fonction à chaque élément d’une liste)

````python
nombres = [1, 2, 3, 4]
carres = list(map(lambda x: x**2, nombres))
print(carres)  # [1, 4, 9, 16]
````

* Avec `filter` (filtrer une liste selon une condition)

````python
pairs = list(filter(lambda x: x % 2 == 0, nombres))
print(pairs)  # [2, 4]
````

* Avec `sorted` (trier selon un critère)

````python
etudiants = [("Alice", 22), ("Bob", 19), ("Charlie", 23)]

# Trier par âge
etudiants_tries = sorted(etudiants, key=lambda x: x[1])
print(etudiants_tries)  
````

# Décorateurs

En Python, les décorateurs sont un moyen flexible de modifier ou d’étendre le comportement de fonctions ou de méthodes, sans changer leur code d’origine.
Un décorateur est essentiellement une fonction qui prend une autre fonction en argument et qui retourne une nouvelle fonction avec des fonctionnalités améliorées.

Les décorateurs sont souvent utilisés dans des contextes comme la journalisation (logging), l’authentification ou la mémorisation (memoization). Ils permettent d’ajouter des fonctionnalités supplémentaires à des fonctions ou méthodes existantes de manière propre et réutilisable.

Utilité :
- Ajouter des fonctionnalités sans modifier le code d’origine
- Réutilisable et modulaire

````python
def decorator(func):
    def wrapper():
        print("Before calling the function.")
        func()
        print("After calling the function.")
    return wrapper

@decorator # Applying the decorator to a function
def greet():
    print("Hello, World!")
greet()
````

#### Décorateurs intégrés :
```
@staticmethod : méthode statique
@classmethod : méthode de classe
@property : transforme une méthode en attribut
```

# Copie profonde

En Python, il est important de comprendre la différence entre **affectation**, **copie superficielle** et **copie profonde**, car elles influencent la manière dont les objets sont manipulés en mémoire.

### 1. Copie simple (= affectation)
Quand on écrit :

````python
a = [1, 2, 3]
b = a
````

* Ici, `a` et `b` pointent vers le **même objet** en mémoire.
* Si on modifie `b`, cela modifie aussi `a`.

Exemple :

````python
b[0] = 99
print(a)  # [99, 2, 3]
print(b)  # [99, 2, 3]
````

### 2. Copie superficielle (`copy.copy`)
On peut utiliser le module `copy` pour créer une nouvelle copie :

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
````

* `b` est une **nouvelle liste**, mais ses éléments **internes** (les sous-listes) pointent vers les mêmes objets que ceux de `a`.
* Modifier une sous-liste affectera donc les deux.

Exemple :

````python
b[0][0] = 99
print(a)  # [[99, 2], [3, 4]]
print(b)  # [[99, 2], [3, 4]]
````

### 3. Copie profonde (`copy.deepcopy`)

Avec `deepcopy`, on duplique **toute la structure**, y compris les objets imbriqués :

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
````

* `a` et `b` sont totalement indépendants.
* Modifier `b` n’affecte pas `a`.

Exemple :

````python
b[0][0] = 99
print(a)  # [[1, 2], [3, 4]]
print(b)  # [[99, 2], [3, 4]]
````

# Gestion d'erreurs

Les erreurs (exceptions) interrompent le programme si elles ne sont pas gérées. Pour éviter l’arrêt brutal du programme, Python permet de gérer ces erreurs avec les blocs `try`, `except`, `else`, `finally`.

### 1. Le bloc `try` / `except`
Permet de tester une portion de code et de capturer les erreurs spécifiques.

````python
try:
    x = 10 / 0
except ZeroDivisionError:
    print('Division par zéro interdite')
````

### 2. Le bloc `else`

Le bloc `else` est exécuté uniquement **si aucune erreur n’est survenue** dans le `try`.

````python
try:
    f = open("data.txt")
except FileNotFoundError:
    print("Fichier introuvable")
else:
    print("Fichier ouvert avec succès")
````

### 3. Le bloc `finally`

Le bloc `finally` est **toujours exécuté**, qu’il y ait une erreur ou non.
Il est très utile pour **libérer des ressources** (fermer un fichier, libérer une connexion réseau, etc.).


````python
try:
    f = open("fichier.txt")
finally:
    f.close()
````

### 4. Lever une erreur manuellement (`raise`)

On peut déclencher volontairement une erreur avec `raise`, par exemple pour signaler une mauvaise entrée utilisateur.

````python
age = -5
if age < 0:
    raise ValueError("Âge invalide")
````

# Multithreading et Multiprocessing

**Qu’est-ce qu’un processus en Python ?**
En informatique, un processus est une instance d’un programme informatique en cours d’exécution. Tout processus comporte 3 composants de base :
Un programme exécutable.
Les données associées nécessaires au programme (variables, espace de travail, tampons, etc.).
Le contexte d’exécution du programme (état du processus).

**Qu’est-ce qu’un thread?**
Un thread est une entité au sein d’un processus qui peut être planifiée pour exécution.
C’est-à-dire une séquence d’instructions dans un programme qui peut être exécutée indépendamment du reste du code.
Un thread contient toutes ses informations dans une Table de Contrôle de Thread (TCB – Thread Control Block).

### multithreading vs multiprocessing
- **Multithreading** : plusieurs tâches légères (threads) dans un même processus (pratique pour les entrées/sorties).
- **Multiprocessing** : plusieurs processus indépendants (idéal pour les calculs lourds).

#### Multithreading
Partage de mémoire : 
- Les threads accèdent aux mêmes variables → plus simple mais attention aux conflits (verrous nécessaires).

Limite en Python : 
- À cause du GIL (Global Interpreter Lock), un seul thread exécute du code Python à la fois.

Quand l’utiliser ? 
- Tâches I/O bound (attente : fichiers, réseau, base de données).


````python
import threading

def worker():
    print('Travail dans un thread')

thread = threading.Thread(target=worker)
thread.start()
thread.join()
````

#### Multiprocessing
Mémoire séparée : 
- Chaque processus a son espace mémoire → pas de GIL.

Quand l’utiliser ?
- Tâches CPU bound (calculs lourds : IA, traitement d’images).


````python
import multiprocessing

def carre(x):
    return x * x

if __name__ == "__main__":  # À rouler dans un script
    with multiprocessing.Pool(4) as p:
        print(p.map(carre, [1, 2, 3, 4]))
````
